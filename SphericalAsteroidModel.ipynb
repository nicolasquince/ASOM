```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
G = 6.67430e-11  # Gravitational constant, m^3/kg/s^2
pi = np.pi

# Mass error analysis parameters
R = 25  # Radius of the asteroid in meters
rho = 2000  # Density in kg/m^3
h = 1e3  # Closest approach distance in meters for this analysis
V = 1e3  # Initial velocity considered

# Compute the mass and ΔV of the asteroid
M = (4/3) * pi * R**3 * rho         # Mass of the spherical asteroid
DeltaV = G*M*pi/(h*V)               # DeltaV given by the initial velocity
deltaM_deltaDeltaV = V*h/(G*np.pi)  # Sensitivity in mass computation wrt ΔV

(M,DeltaV,deltaM_deltaDeltaV)       # Display of the interesting quantities
```
